```python
import pandas as pd

# ---------------------------------------------------------
# OPCIÓN A: Si tu archivo grande es un CSV
# ---------------------------------------------------------
ruta_csv_origen = 'NYC_BikeShare.csv'
ruta_csv_destino = f'./muestras/{ruta_csv_origen}_muestra_500_filas.csv'

# Usamos nrows para leer solo las primeras 100 filas rápidamente
df_csv = pd.read_csv(ruta_csv_origen, nrows=500)

# Guardamos la muestra
df_csv.to_csv(ruta_csv_destino, index=False)
print("¡Muestra CSV guardada con éxito!")
```

```python
# ---------------------------------------------------------
# OPCIÓN B: Si tu archivo grande es un Excel (.xlsx)
# ---------------------------------------------------------
ruta_excel_origen = '.'
ruta_excel_destino = f'{ruta_excel_origen}_muestra_100_filas.xlsx'

# Leemos las primeras 100 filas de la hoja principal
df_excel = pd.read_excel(ruta_excel_origen, nrows=500)

# Guardamos la muestra
df_excel.to_excel(ruta_excel_destino, index=False)
print("¡Muestra Excel guardada con éxito!")
```
